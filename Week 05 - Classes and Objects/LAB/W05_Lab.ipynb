### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
class Record:
    """Represent a record with a label, value and limit."""

    def __init__(self, label, value, limit):
        """Store the label, value and limit in the object."""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a string containing the label, value and limit."""
        return f"{self.label}: {self.value}/{self.limit}"

    def status(self):
        """Return the status based on the percentage of the limit."""
        percent = (self.value / self.limit) * 100

        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"


r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

print(r1.status())
print(r2.status())
print(r3.status())
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
class Record:
    """Represent a record with a label, value and limit."""

    def __init__(self, label, value, limit):
        """Store the label, value and limit in the object."""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a string containing the label, value and limit."""
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value / self.limit) * 100

    def status(self):
        """Return the status using the percentage method."""
        percent = self.percent()

        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"


r = Record("R-001", 87.0, 100.0)

print(r.percent())
print(r.status())
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
class Checker:
    """Count how many checked values exceed their limits."""

    def __init__(self):
        """Set the over-limit counter to zero."""
        self.over_limit_count = 0

    def check(self, value, limit):
        """Increase the counter when the value exceeds the limit."""
        if value > limit:
            self.over_limit_count += 1


checker = Checker()

checker.check(87, 100)
checker.check(120, 100)
checker.check(130, 100)

print(checker.over_limit_count)
````

### D10. A class-level counter

1. Define a new, small class `Counted`. Directly under the `class` line (not inside `__init__`),
   add `count = 0`.

3. Create three `Counted` objects, then print `Counted.count`.

````python
class Counted:
    """Count how many objects are created."""

    # The count belongs to the class, so it is shared by all its objects.
    count = 0

    def __init__(self):
        """Increase the shared counter when an object is created."""
        Counted.count += 1


c1 = Counted()
c2 = Counted()
c3 = Counted()

print(Counted.count)
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
def load_records(path):
    """Read a file and return a list of Record objects."""
    records = []

    with open(path) as f:
        for line in f:
            label, value, limit = line.strip().split(",")

            value = float(value)
            limit = float(limit)

            record = Record(label, value, limit)
            records.append(record)

    return records


records = load_records("records_sample.txt")

for record in records:
    print(f"{record.label} {record.status()}")
````

### C1. Count the records over the limit

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
class Record:
    """Represent a record with a label, value and limit."""

    def __init__(self, label, value, limit):
        """Store the label, value and limit in the object."""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a string containing the label, value and limit."""
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value / self.limit) * 100

    def status(self):
        """Return the status using the percentage method."""
        percent = self.percent()

        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"

    def is_over(self):
        """Return True if the value reaches or exceeds the limit."""
        return self.value >= self.limit


def load_records(path):
    """Read a file and return a list of Record objects."""
    records = []

    with open(path) as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            record = Record(label, float(value), float(limit))
            records.append(record)

    return records


records = load_records("records_sample.txt")

over_limit_count = 0

for record in records:
    if record.is_over():
        over_limit_count += 1

print(f"{over_limit_count} of {len(records)} records are OVER LIMIT")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
class Account:
    """Represent an account with a username and failed login count."""

    def __init__(self, username):
        """Store the username and set failed logins to zero."""
        self.username = username
        self.failed_logins = 0

    def fail(self):
        """Increase the failed login count by one."""
        self.failed_logins += 1

    def locked(self):
        """Return True if there are three or more failed logins."""
        return self.failed_logins >= 3


anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())

ben = Account("ben")
ben.fail()
print(ben.locked())
````

1. **Classes** - A class describes a type of object. `__init__`sets its starting attributes when a new object is created.

````python
class Server:
    def __init__(self, name):
        self.name = name

server = Server("srv-01")
print(server.name)   
````

2. `self` - refers to the particular object that is using the method.

````python
class Server:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)

server = Server("srv-01")
server.show_name()   
````

3. **Attributes** - store data on an object. I access them using a dot.

````python
class Server:
    def __init__(self, name):
        self.name = name

server = Server("srv-01")

print(server.name)        
server.name = "srv-main"
print(server.name)        
````

4. **Objects are separate** - Each object can have its own attributes. Changing one object's own data does not change another object's data.

````python
first = Server("srv-01")
second = Server("srv-02")

first.name = "main"

print(first.name)    
print(second.name)   
````

5. **Methods** - A method is a function inside a class. I call it through an object, and it can call another method using `self`.

````python
class Server:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)

    def report(self):
        self.show_name()

server = Server("srv-01")
server.report() 
````

6.**Methods can change their own object** - A method can update an attribute so the object remembers a change between calls.

````python
class Counter:
    def __init__(self):
        self.count = 0

    def increase(self):
        self.count += 1

counter = Counter()
counter.increase()
counter.increase()

print(counter.count) 
````

7.**Class-level attributes** - A class attribute belongs to the class and is shared by its objects, unless an object defines its own attribute with the same name.

````python
class Server:
    category = "Network"

first = Server()
second = Server()

Server.category = "Infrastructure"

print(first.category)    # Infrastructure
print(second.category)   # Infrastructure
````

8. **A list of objects** -  can store objects. I can load their data from a file, create the objects, and call a method on each one.

````python
with open("servers.txt", "w") as f:
    f.write("srv-01\n")
    f.write("srv-02\n")

servers = []

with open("servers.txt") as f:
    for line in f:
        name = line.strip()
        servers.append(Server(name))

for server in servers:
    server.report()
````
