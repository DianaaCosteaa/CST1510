Write them in your own words, with a short example for each. Do not copy from the slides.

1. Comparisons check how two values relate and give True or False. = assigns a value; == checks whether two values are equal.

````python
x = 5           # Store 5 in x

print(x == 5)   # True 
print(x != 5)   # False 
print(x < 5)    # False 
print(x > 5)    # False 
print(x <= 5)   # True 
print(x >= 5)   # True
````

2. Combining conditions  `and` needs both conditions to be `True`. `or` needs at least `one to be True`. `not` reverses True and False.

````python
age = 20

print(age >= 18 and age <= 65)   # True
print(age < 18 or age > 65)      # False
print(not age == 20)            # False
````

3. Decisions - Python checks the conditions from top to bottom and runs the first True branch. If none is True, it runs else.

````python
score = 50

if score >= 70:
    print("Excellent")
elif score >= 40:
    print("Pass")
else:
    print("Fail")
````

4. Indentation - Indentation shows which instructions belong inside an if or a loop. I use four spaces for each level.

````python
age = 16

if age >= 18:
    print("Adult")

print("Finished")
````

5. while loops - A while loop repeats while its condition is True. To finish, the condition must become False or the loop must reach break. `while True` repeats until something stops it. `break` exits the loop.

````python
number = 1

while number <= 3:
    print(number)
    number += 1
````

````python
while True:
    answer = input("Type stop to finish: ")

    if answer == "stop":
        break
````

`continue` skips the rest of the current repetition and starts the next one.

````python
number = 0

while number < 5:
    number += 1

    if number == 4:
        continue

    print(number)
````

6. `for` loops - A for loop goes through a sequence, one item at a time. With range(), the stop value is not included.

````python
for number in range(3):                   # range(stop)
    print(number)
````

````python
for number in range(2, 7, 2):          # range(start, stop, step)   
    print(number)
````

`enumerate()` gives a position and a value together.

````python
for position, character in enumerate("ABC", start=5):
    print(position, character)
````

7. `for` or `while`?

I use `for` for a fixed number of repetitions or to go through items. I use `while` when repetition depends on a condition.

````python
for count in range(3):
    name = input("Enter a name: ")
    print(name)
````

````python
password = ""

while password != "letmein":
    password = input("Enter the password: ")

print("Welcome")
````
